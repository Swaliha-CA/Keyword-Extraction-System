```python
print("KEYWORD EXTRACTION SYSTEM")
print("-" * 35)

print("Enter your text below.")
print("Type END on a new line when finished.\n")

lines = []

while True:
    line = input()

    if line.strip().upper() == "END":
        break

    lines.append(line)

user_text = "\n".join(lines)

if len(user_text.split()) < 3:
    print("Please enter at least three words.")
else:
    print("\nText received successfully!")
    print("Total words:", len(user_text.split()))
```
